```python
# Q: How does KNN work from scratch?

# KNN Pipeline in model :
# New sample
# → Calculate distances
# → Sort by distance
# → Select K nearest neighbors
# → Get their labels
# → Vote (classification) / Average (regression)
# → Prediction

# KNN Engineering Checklist:
# 1. Prepare clean data
# 2. Select useful features
# 3. Scale features when needed
# 4. Choose a distance metric
# 5. Choose K
# 6. Find K nearest neighbors
# 7. Predict using vote/average
# 8. Validate performance
```

```python
# Q: How do interpolation and extrapolation relate to KNN?

# Interpolation:
# Predict inside the region covered by training data
# KNN works naturally because nearby data points exist

# Extrapolation:
# Predict outside the region covered by training data
# KNN has weaker evidence because nearby points may not exist

# KNN doesn't learn the road ahead; it looks at the nearest points around you
```
