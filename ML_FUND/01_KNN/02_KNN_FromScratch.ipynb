```python
import numpy as np
from collections import Counter

class ScratchKNN:

    def __init__(self, k: int = 3):
        """
        Initialize the KNN classifier.
        
        Parameters:
        k(int): Number of nearest neighbors to consider.
        """
        self.k = k
        self.X_train = None
        self.y_train = None

    def fit(self, X_train, y_train) -> None:
        """
        Store the training data into the classifier.
        
        Parameters:
        X_train: array-like of shape (n_samples, n_features)
        y_train: array-like of shape (n_samples,)
        """
        self.X_train = np.array(X_train)
        self.y_train = np.array(y_train)

    def _compute_distances(self, X_test: np.ndarray) -> np.ndarray:
        """
        Compute Euclidean distance between each test sample and all training samples 
        using vectorized NumPy operations.
        
        Parameters:
        X_test: numpy array of shape (n_test_samples, n_features)
        
        Returns:
        distances: numpy array of shape (n_test_samples, n_train_samples)
        """
        # Using broadcasting to compute Euclidean distance efficiently:
        # sqrt(sum((x - y)^2))
        distances = np.sqrt(np.sum((X_test[:, np.newaxis, :] - self.X_train[np.newaxis, :, :]) ** 2, axis=2))
        return distances

    def predict(self, X_test) -> np.ndarray:
        """
        Predict the class labels for the provided test data.
        
        Parameters:
        X_test: array-like of shape (n_test_samples, n_features)
        
        Returns:
        predictions: numpy array of predicted labels
        """
        X_test = np.array(X_test)
        
        # Step 1: Compute distances to all training points
        distances = self._compute_distances(X_test)
        
        predictions = []
        for dist_row in distances:
            # Step 2: Find the indices of the K smallest distances
            k_indices = np.argsort(dist_row)[:self.k]
            
            # Step 3: Gather the corresponding labels of the K nearest neighbors
            k_nearest_labels = self.y_train[k_indices]
            
            # Step 4: Perform majority vote to find the most common class label
            most_common = Counter(k_nearest_labels).most_common(1)
            predictions.append(most_common[0][0])
            
        return np.array(predictions)
```

```python
if __name__ == "__main__":
    # Sample training data (features and binary labels)
    X_train = [
        [1.0, 2.0],
        [1.5, 1.8],
        [8.0, 8.0],
        [8.5, 9.0]
    ]
    y_train = [0, 0, 1, 1]

    # Initialize and train our custom KNN classifier (k=3)
    knn = ScratchKNN(k=3)
    knn.fit(X_train, y_train)

    # Test samples to predict
    X_test = [
        [1.2, 1.5],
        [8.2, 8.5]
    ]

    # Make predictions
    predictions = knn.predict(X_test)
    print("Predictions:", predictions)  # Expected output: [0, 1]
```
