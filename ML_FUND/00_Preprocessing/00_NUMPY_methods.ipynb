```python
import numpy as np
```

```python
#===================================================================================
# Slicing
#===================================================================================


# Slicing in 1_D arrays
# [start:stop:step]
arr = np.array([1, 2, 3, 4, 5])
print(arr[1:4])  # Output: [2 3 4]
print(arr[2:-1])  # Output: [3 4 5]
print(arr[::2])  # Output: [1 3 5]
print(arr[::-1])  # Output: [5 4 3 2 1]


# Slicing in 2_D arrays
```
