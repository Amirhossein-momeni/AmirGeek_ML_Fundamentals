```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#============================================================================================
# Import Dataset
#============================================================================================

# import data
URL = "https://raw.githubusercontent.com/Amirhossein-momeni/AmirGeek_ML_Datasets/refs/heads/main/DataSet/03_flights.csv"
df = pd.read_csv(URL)
df.head(5)
```

```python
#============================================================================================
# EDA with Python (Exploratory Data Analysis) - Descriptive Statistics
#============================================================================================

# Trim mean 
from scipy import stats
df_trim_mean = stats.trim_mean(df['arr_delay'], 0.1)
print(f"Trim Mean Arrival Delay: {df_trim_mean}")
print(60*"=")

# Median 
df_median = df['arr_delay'].median()
print(f"Median Arrival Delay: {df_median}")
print(60*"=")

# Mode
df_mode = df['arr_delay'].mode()
print(f"Mode Arrival Delay: {df_mode.iloc[0]}")
print(60*"=")

# STD
df_STD = df['arr_delay'].std()
print(f"STD Arrival Delay: {df_STD}")
print(60*"=")
```
