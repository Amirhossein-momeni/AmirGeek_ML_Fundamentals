```python
import pandas as  pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#==============================================================================
# Reading DataSet
#==============================================================================

# Load dataset
dataset = sns.load_dataset("titanic")
```

```python
#==============================================================================
# preprocessing
#==============================================================================

# drop deck column because it has too many missing values
dataset = dataset.drop(columns=['deck'], errors='ignore')

# drop embark_town column because it is redundant with embarked
dataset = dataset.drop(columns=['embark_town'], errors='ignore')

# drop rows with missing values in age and embarked columns
dataset = dataset.dropna(subset=['age', 'embarked'])
```
