```python
# What are bias and variance, and how do I diagnose them?

# Bias-Variance:

# High Bias
# → Model too simple
# → Misses important patterns
# → Underfitting
# → Train: bad / Validation: bad

# High Variance
# → Model too sensitive to training data
# → Learns noise/details
# → Overfitting
# → Train: good / Validation: bad

# Model Complexity or Regularization:
# Too simple → High Bias ( train: bad / validation: bad )
# Balanced    → Good Generalization ( train: good / validation: good ) 
# Too complex → High Variance ( train: good / validation: bad )

# KNN example:
# K ↓ → Variance ↑ → Overfitting risk ↑
# K ↑ → Bias ↑ → Underfitting risk ↑

# Key:
# Bias = too simple
# Variance = too sensitive
```
